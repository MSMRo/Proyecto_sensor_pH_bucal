# Simulador de regresion lineal usando ScatterWidget

```python
from drawdata import ScatterWidget
```

## Generación de PH sintético

```python
widget = ScatterWidget()
widget
```

```python
import pandas as pd

df = widget.data_as_pandas
df
```

```python
data = df[['x','y']].to_numpy()
X = data[:,0].reshape(-1,1)
y = data[:,1]
```

```python
import matplotlib.pyplot as plt

# Plot
plt.scatter(X, y, label="Data")           # original points
plt.xlabel("PH obtained")
plt.ylabel("PH real")
plt.grid(ls=":")
plt.title("Linear Regression Fit")
plt.legend()
plt.show()
```

## Creación de modelo de regresión lineal

```python
from sklearn import linear_model
model = linear_model.LinearRegression()

model.fit(X,y)

# Parameters
print(model.coef_)      # slope(s)
print(model.intercept_) # bias term
```

### Obtención de score

```python
# Predictions
y_pred = model.predict(X)
```

```python
from sklearn.metrics import r2_score, mean_squared_error

r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)

print(f"R²: {r2:.4f}")
print(f"MSE: {mse:.4f}")
```

```python
# Plot
plt.scatter(X, y, label="Data")           # original points
plt.plot(X, y_pred, label="Regression", color="red")   # fitted line
plt.grid(ls=":")
plt.xlabel("PH obtained")
plt.ylabel("PH real")
plt.title("Linear Regression Fit")
plt.legend()
plt.show()
```
